```python
s = "data engineering data python"
words = s.split()

freq = {}

for word in words:
    if word in freq:
        freq[word] += 1
    else:
        freq[word] = 1

print(freq)
```

```python
data = [10,12,12,13,14,15,15]

unq = []
dup = []

for i in data:
    if i not in unq:
        unq.append(i)
    elif i not in dup:
        dup.append(i)
print(dup)
print(unq)
```

```python
numbers = [10, 50, 30, 20]

largest = numbers[0]

for n in numbers:
    if n > largest:  #10 > 10 #50 > 10 #30 >50 #20 >50
        largest = n

print(largest)
```
